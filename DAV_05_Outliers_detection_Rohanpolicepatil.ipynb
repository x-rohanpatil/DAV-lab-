```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore

data = {
     "Stduent":["A","B","C","D","E","F","G","H","I","J","K","L","M","N","O"],
     "Study_hours":[2,3,4,5,4,6,5,3,4,5,6,4,5,30,3]

}
df = pd.DataFrame(data)
print("Original Dataset")
print(df)
```

```python
df["Study_hours"].describe()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df["Study_hours"])
plt.title("Study Hours - Boxplot")
plt.xlabel("Study Hours")

plt.show()
```

**INTER QUANTILE** **RANGE**
[IQR] METHOD

```python
Q1 = df["Study_hours"].quantile(0.25)
Q3 = df["Study_hours"].quantile(0.75)

print("Q1:",Q1)
print("Q3:",Q3)
```

```python
IQR = Q3 - Q1
print("IQR:",IQR)
```

```python
lower_bound = Q1 -1.5*IQR
upper_bound = Q3 +1.5*IQR

print("Lower Bound:",lower_bound)
print("Upper Bound:",upper_bound)
```

```python
outliers_iqr = df[
    (df["Study_hours"] < lower_bound) | (df["Study_hours"] > upper_bound)
]

print("Outliers (IQR Method):")
display(outliers_iqr)
```

```python
outliers_iqr = df[
    (df["Study_hours"] >= lower_bound) & (df["Study_hours"] <= upper_bound)
]

print("Non outliers")
display(outliers_iqr)
```

```python
df["z_score"] = zscore(df["Study_hours"])
print("Dataset with z_scores")
print(df)
```

```python
z_outliers = df[
    (df["z_score"]>3)|
    (df["z_score"]<-3)
]

print("Z-score outliers")
display(z_outliers)
```

```python
df_without_outliers = df[
    (df["Study_hours"]>= lower_bound)&
    (df["Study_hours"]<= upper_bound)
]

print("Dataset without outliers")
display(df_without_outliers)
```
