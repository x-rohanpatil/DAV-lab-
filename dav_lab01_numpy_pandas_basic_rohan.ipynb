```python
import numpy as np
import pandas as  pd

print("Numpy version:",np.__version__)
print("Pandas version:",pd.__version__)

```

```python
marks = np.array([23,45,56,67,45])

print("Marks: ",marks)
print("Type: ",type(marks))
```

```python
print("Number of dimensions: ",marks.ndim)
print("Shape of array: ",marks.shape)
print("Number 0f elements: ",marks.size)
```

```python
marks = np.array([23,45,56,67,45])
print("oriignal: ",marks)
print("oriignal: ",marks+5)
print("multiply by 2: ",marks*2)
print("divide by 2 :",marks/2)
print("square: ",marks**2)
```

```python
print("sum:",marks.sum())
print("min:",marks.min())
print("max:",marks.max())
print("mean:",marks.mean())
```

```python
marks = np.array([23,45,56,67,45])

print("First mark: ",marks[0])
print("second number : ",marks[1])
print("last number :",marks[-1])
```

```python
marks = np.array([23,45,56,67,45])
print("index 1 to 3:",marks[1:4])
print("from index 2 :",marks[2:])
print("first three :",marks[:3])
print("every second element: ",marks[::2])
```

```python


student_marks = np.array([[10,11,12],[14,15,16],[17,18,19]])

print(student_marks)
print("shape: ",student_marks.shape)

print("Mean of all values: ",student_marks.mean())
print("column wise mean: ",student_marks.mean(axis=0))
print("Row wise mean: ",student_marks.mean(axis=1))
```
